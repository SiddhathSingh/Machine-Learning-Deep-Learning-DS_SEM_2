```python
df = df.replace([np.inf, -np.inf], np.nan)

num_cols = ["Salary Estimate","Rating","Revenue","min_salary","max_salary","avg_salary","age"]

for col in num_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")

for col in num_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    df = df[(df[col].isna()) |
        ((df[col] >= lower) & (df[col] <= upper)) ]

print("After removing outliers:", df.shape)
```

```python

salaryData = df.dropna(subset=["avg_salary"]).copy()

columns_to_drop = [
    "avg_salary",
    "Salary Estimate",   
    "min_salary",        
    "max_salary",        
    "Job Description",  
    "Competitors"        
]
X = df.drop(columns=[c for c in columns_to_drop if c in df.columns])
y = df["avg_salary"].astype(float)

# Remove duplicate/identifier-like columns if present; retain useful job/company categories.
for col in ["a", "Company Name"]:
    if col in X.columns:
        X = X.drop(columns=col)

print("Features:", X.shape)
print("avg_salary", y.shape)
X.head()
```

# Split Data

```python
X = df.select_dtypes(include=np.number).drop(['avg_salary'], axis=1)
y = df[['avg_salary']]

model_data = pd.concat([X, y], axis=1)
model_data = model_data.dropna()

X = model_data.drop(['avg_salary'], axis=1)
y = model_data[['avg_salary']]

print(X.shape)
print(y.shape)
```

# Train Test split

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.30, random_state=1)
```

# First linear model

```python
regression_model = LinearRegression()
regression_model.fit(X_train, y_train)
```

# Intercept

```python
intercept = regression_model.intercept_[0]
print("The intercept for our model is {}".format(intercept))
```

# Training R^2

```python
regression_model.score(X_train, y_train)
```

# Testing r^2

```python
regression_model.score(X_test, y_test)
```

# Adding interaction terms

```python
from sklearn.preprocessing import PolynomialFeatures
from sklearn import linear_model

poly = PolynomialFeatures(
    degree=2,
    interaction_only=True
)

X_train2 = poly.fit_transform(X_train)

X_test2 = poly.transform(X_test)
```

# Fit linear model

```python
poly_clf = linear_model.LinearRegression()
poly_clf.fit(X_train2, y_train)
```

# Prediction

```python
y_pred = poly_clf.predict(X_test2)
print(y_pred[:10])
```

# Training R^2

```python
print(poly_clf.score(X_train2, y_train))
```

# Testing R^2

```python
print(poly_clf.score(X_test2, y_test))
```
